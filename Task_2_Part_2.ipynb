```python
import pandas as pd
from transformers import pipeline
from sklearn.metrics import classification_report
```

```python
# ======================================================
# 1. PROBLEM STATEMENT
# ======================================================
print("📌 STEP 1: PROBLEM STATEMENT")
print("Objective: Automatically tag support tickets into categories using a Zero-Shot Classification LLM.")
print("We will use the 'facebook/bart-large-mnli' model.")
print("-" * 50)
```

```python
# ======================================================
# 2. CREATE DUMMY DATASET (Fastest method)
# ======================================================
print("📌 STEP 2: CREATING DATASET")
# Instead of downloading a massive file, we create a representative dataset here.
data = {
    "text": [
        "My internet is not working and the light is blinking red.",
        "I want to return the shoes I bought yesterday, they are too small.",
        "Can you tell me if my payment went through? I didn't get a receipt.",
        "The app keeps crashing when I try to upload a photo.",
        "I forgot my password and cannot login to my account.",
        "Where is my order? It was supposed to arrive on Monday."
    ],
    "actual_category": [
        "Technical Issue",
        "Returns & Refunds",
        "Billing & Payments",
        "Technical Issue",
        "Account Access",
        "Shipping & Delivery"
    ]
}

df = pd.DataFrame(data)
print("✅ Dataset Created!")
print(df.head())
print("-" * 50)
```

```python
# ======================================================
# 3. LOAD LLM MODEL (Zero-Shot Classifier)
# ======================================================
print("📌 STEP 3: LOADING LLM MODEL")
print("⏳ Downloading model (this may take a moment)...")

# We use the BART Large MNLI model which is excellent for zero-shot classification
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

print("✅ Model Loaded Successfully!")
print("-" * 50)
```

```python
# ======================================================
# 4. PREDICT TAGS
# ======================================================
print("📌 STEP 4: PREDICTING TAGS")

# Define our candidate labels (The categories we want the AI to choose from)
candidate_labels = ["Technical Issue", "Returns & Refunds", "Billing & Payments", "Account Access", "Shipping & Delivery"]

# Function to predict
def predict_tag(text):
    result = classifier(text, candidate_labels)
    return result['labels'][0] # Returns the highest probability tag

# Apply to the dataframe
print("⏳ Classifying tickets...")
df['predicted_category'] = df['text'].apply(predict_tag)

print("✅ Classification Complete!")
print("-" * 50)
```

```python
# ======================================================
# 5. EVALUATION & SUMMARY
# ======================================================
print("📌 STEP 5: RESULTS & SUMMARY")

# Show results
print(df[['text', 'actual_category', 'predicted_category']])

# Calculate simple accuracy
accuracy = (df['actual_category'] == df['predicted_category']).mean()
print(f"\n🎯 Zero-Shot Accuracy: {accuracy*100:.2f}%")

print("\nFinal Summary:")
print("1. Used 'facebook/bart-large-mnli' for Zero-Shot Classification.")
print("2. No training data was required (Zero-Shot Learning).")
print(f"3. The model successfully tagged tickets with {accuracy*100:.0f}% accuracy.")
```
